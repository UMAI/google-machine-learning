```python
import pandas as pd

pd.options.display.max_rows = 10
pd.options.display.float_format = "{:.1f}".format
```

```python
training_df = pd.read_csv(filepath_or_buffer="https://download.mlcc.google.com/mledu-datasets/california_housing_train.csv")
```

```python
training_df.describe()
```

```python
for feature in training_df.columns:
    if (
        training_df[feature].min() < training_df[feature].mean() - 3 * training_df[feature].std()
        or training_df[feature].max() > training_df[feature].mean() + 3 * training_df[feature].std()
    ):
        print(f"Column {feature} contains outliers (> 3 std from mean)")
    
```

```python
is_outlier_df = (training_df - training_df.mean()).abs() > 3 * training_df.std()
is_outlier_df.loc[:, is_outlier_df.any()].columns
```

```python
low_deltas = (training_df.quantile(.25) - training_df.min()).abs()
high_deltas = (training_df.quantile(.75) - training_df.max()).abs()
outlier_scale = high_deltas / low_deltas > 4
outlier_scale.index[outlier_scale == True]
```
